```python
def dinner_split():
    print("=== 聚餐分帳計算器 ===")
    try:
        total_amount = float(input("請輸入總金額 (元): "))
        service_rate = float(input("請輸入服務費百分比 (例如 10 代表 10%，無則輸入 0): "))
        people = int(input("請輸入分攤人數: "))

        if total_amount < 0 or service_rate < 0 or people <= 0:
            print("輸入數值錯誤：金額與服務費不可為負數，且人數必須大於 0。")
            return

        service_fee = total_amount * (service_rate / 100)
        final_total = total_amount + service_fee
        per_person = final_total / people

        print("\n--- 計算結果 ---")
        print(f"原始金額：{total_amount:.0f} 元")
        print(f"服務費 ({service_rate:.0f}%)：{service_fee:.0f} 元")
        print(f"含服務費總金額：{final_total:.0f} 元")
        print(f"每人應付金額：{per_person:.1f} 元")

    except ValueError:
        print("輸入格式錯誤，請輸入有效的數字！")

dinner_split()
```
